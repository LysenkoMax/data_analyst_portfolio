# Анализ LTV по штатам Бразилии

1. Подготавливаемся к работе и импортируем нужные библиотеки

```python
%matplotlib inline 
# Устанавливаем для отображения графиков библиотеки Matplotlib в Jupyter Notebook
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

2. Загружаем данные из CSV

```python
df = pd.read_csv('state_ltv.csv')
```

3. Посмотрим на данные

```python
print("Первые 5 строк данных:")
print(df.head())
```

4. Построение визаулизации по данным

```python
# Настройка размера
plt.figure(figsize=(15, 8)) 

# Строим график
plot = sns.barplot(data=df, x='customer_state', y='avg_ltv', 
                   hue='customer_state', palette='viridis', legend=False)

# Названия и подписи
plt.title('Средний LTV клиента в разрезе штатов Бразилии', fontsize=16)
plt.xlabel('Штат', fontsize=12)
plt.ylabel('Средний LTV (BRL)', fontsize=12)
plt.xticks(rotation=45)

# Цифры над столбиками
for p in plot.patches:
    plot.annotate(format(p.get_height(), '.1f'), 
                   (p.get_x() + p.get_width() / 2., p.get_height()), 
                   ha = 'center', va = 'center', 
                   xytext = (0, 9), 
                   textcoords = 'offset points')

# Сохранение и показ
plt.savefig('ltv_by_state.png', bbox_inches='tight')
plt.show()
```
